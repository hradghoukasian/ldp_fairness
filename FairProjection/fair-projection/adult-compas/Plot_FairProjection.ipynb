# Results for COMPAS

```python
import pickle
import numpy as np
from scipy.stats import t


# Load the pickle file
file_path = './compas-mp.pkl'

with open(file_path, 'rb') as f:
    data = pickle.load(f)

# Extract dictionaries from the loaded file
gbm_kl_meo      = data['gbm_kl_meo']
gbm_kl_sp       = data['gbm_kl_sp']
gbm_kl_sp_ldp   = data['gbm_kl_sp_ldp']
gbm_kl_meo_ldp  = data['gbm_kl_meo_ldp']
tolerance       = data['tolerance']


# CI helper
def mean_ci(values, confidence=0.95):
    values = np.array(values).flatten()
    n = len(values)
    mean = np.mean(values)
    std = np.std(values, ddof=1)

    t_val = t.ppf((1 + confidence) / 2., n - 1)
    se = std / np.sqrt(n)
    ci = t_val * se
    return mean, ci



# Process everything
save = {
    'gbm_kl_meo': gbm_kl_meo,
    'gbm_kl_sp': gbm_kl_sp,
    'gbm_kl_sp_ldp': gbm_kl_sp_ldp,
    'gbm_kl_meo_ldp': gbm_kl_meo_ldp,
    'tolerance': tolerance
}

keys_to_process = ['gbm_kl_meo', 'gbm_kl_sp', 'gbm_kl_sp_ldp', 'gbm_kl_meo_ldp']

for key in keys_to_process:
    print(f"\n=== Results for {key} ===")
    data_block = save[key]

    for metric in ['sp', 'meo_abs', 'acc']:
        arr = np.array(data_block[metric]).flatten()
        mean, ci = mean_ci(arr)
        print(f"{metric.upper()}: {mean:.6f} ± {ci:.6f}")
```
